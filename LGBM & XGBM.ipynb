## Building Predictive Models:

1.	Split the preprocessed dataset into training and testing sets.

2.	Choose appropriate evaluation metrics (e.g., accuracy, precision, recall, F1-score) for model evaluation.

3.	Build predictive models using LightGBM and XGBoost algorithms.

4.	Train the models on the training set and evaluate their performance on the testing set.

5.	Use techniques like cross-validation and hyperparameter tuning to optimize model performance.


```python
from sklearn.model_selection import train_test_split
x=df.drop(['survived_1'],axis=1)
y=df['survived_1']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# pip install xgboost
```

```python
# pip install lightgbm
```

```python
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import accuracy_score,precision_score
```

```python
xgb=XGBClassifier(n_estimator=100,learning_rate=0.1,max_dept=5,random_state=42)
xgb.fit(x_train,y_train)
y_pred1=xgb.predict(x_test)
```

```python
print(f'Accuracy Score {(accuracy_score(y_test,y_pred1)):.2f}')
print(f'Precision Score{precision_score(y_test,y_pred1):.2f}')
```

```python
lgb=LGBMClassifier(n_estimator=100,learning_rate=0.1,max_dept=5,random_state=42)
lgb.fit(x_train,y_train)
y_pred2=lgb.predict(x_test)
```

```python
print(f'Accuracy Score {(accuracy_score(y_test,y_pred2)):.2f}')
print(f'Precision Score{precision_score(y_test,y_pred2):.2f}')
```

```python
# This is for XGboost model
from sklearn.model_selection import cross_val_score,GridSearchCV,StratifiedKFold
import numpy as np
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
cv_scores=cross_val_score(xgb,x,y,cv=cv,scoring='accuracy')
print('Cross Validation',cv_scores)
print('Mean CV Accuracy',np.mean(cv_scores))
```

```python
# This is for LGboost model
cv_scores2=cross_val_score(lgb,x,y,cv=cv,scoring='accuracy')
print('Cross Validation',cv_scores2)
print('Mean CV Accuracy',np.mean(cv_scores2))
```

```python
#Hyperparameter tuning for (Gridsearchcv)
# For XGboost
param_grid_xgb={'n_estimators':[100,200,300],
            'max_depth':[5,3,5],
            'learning_rate':[0.1,0.2,0.1]}

grid_xgb=GridSearchCV(param_grid=param_grid_xgb,
                      estimator=XGBClassifier(),
                      cv=5)

grid_xgb.fit(x_train,y_train)

print('Best Parameter for XGB',grid_xgb.best_params_)
print('Best CV Score',grid_xgb.best_score_)
```

```python
# for LGboost
param_grid_lgb={'n_estimators':[100,200,300],
            'max_depth':[5,3,5],
            'learning_rate':[0.1,0.2,0.1]}

grid_lgb=GridSearchCV(param_grid=param_grid_lgb,
                      estimator=LGBMClassifier(),
                      cv=5)

grid_lgb.fit(x_train,y_train)

print('Best Parameter for LGB',grid_lgb.best_params_)
print('Best CV Score',grid_lgb.best_score_)
```

```python
# For Best Estimators

best_xgb=grid_xgb.best_estimator_
y_pred_xgb=best_xgb.predict(x_test)

best_lgb=grid_lgb.best_estimator_
y_pred_lgb=best_lgb.predict(x_test)

print('Best Accuracy of XGB',accuracy_score(y_test,y_pred_xgb))
print('Best Accuracy of LGB',accuracy_score(y_test,y_pred_lgb))
```

## Comparative Analysis:

1.	Compare the performance metrics (e.g., accuracy, precision, recall) of LightGBM and XGBoost models.

2.	Visualize and interpret the results to identify the strengths and weaknesses of each algorithm.


```python
from sklearn.metrics import f1_score,recall_score,classification_report
xgb_accuracy=accuracy_score(y_test,y_pred1)
xgb_precision=precision_score(y_test,y_pred1)
xgb_f1score=f1_score(y_test,y_pred1)
xgb_recall=recall_score(y_test,y_pred1)

lgb_accuracy=accuracy_score(y_test,y_pred2)
lgb_precision=precision_score(y_test,y_pred2)
lgb_f1score=f1_score(y_test,y_pred2)
lgb_recall=recall_score(y_test,y_pred2)
```

```python
print('Classification Report Of XGB\n',classification_report(y_test,y_pred_xgb))
print('\nClassification Report Of LGB\n',classification_report(y_test,y_pred_lgb))
```

```python
metrics=pd.DataFrame({'Model':['XGBoost','LGBoost'],
                      'Accuracy':[xgb_accuracy,lgb_accuracy],
                      'Precision':[xgb_precision,lgb_precision],
                      'Recall':[xgb_recall,lgb_recall],
                      'F1_Score':[xgb_f1score,lgb_f1score]})
print(metrics)
```

```python
plt.figure(figsize=(8,6))
sns.barplot(data=metrics.melt(id_vars='Model'),x='variable',y='value',hue='Model')
plt.xlabel('Score')
plt.ylabel('Metric')
plt.title('Model')
plt.show()
```
